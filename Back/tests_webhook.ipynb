```python
import pandas as pd
import requests
```

```python
ruta_csv = "Pedidos de Kits 27-1 (Respuestas) - Respuestas de formulario 1(1).csv"
data = pd.read_csv(ruta_csv)
```

```python
data = data.astype(object)
data = data.where(pd.notnull(data), None)
```

```python
fila = data.iloc[0].to_dict()
```

```python
print(fila)
```

```python
url = "http://127.0.0.1:8000/api/v1/pipeline_sheets_kits/sheets"
```

```python
try:
    respuesta = requests.post(url, json=fila)
    print(f"Código de estatus: {respuesta.status_code}")
    print(f"Respuesta del servidor: {respuesta.json()}")
except requests.exceptions.ConnectionError:
    print(f"Error de conexión")
```
